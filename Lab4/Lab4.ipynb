```python
#Question4
def factorial(n):
    if n==0 or n==1:
     return 1
    result = 1
    for i in range(2,n+1):
       result*=i
    return result
print(factorial(5))
```

```python
#Question5
#PART A
import math

def f(x):
    if x==0:
        raise ValueError("X cannot be zero for f(x)=1/x")
    return 1/x
print(f(5))
```

```python
def g(x):
    if x<0:
        raise ValueError("X must be positive")
    return math.sqrt(x)
print(g(4))
```

```python
def h(x):
    if x<=0:
        raise ValueError("X cannot be negative or zero")
    return math.log(x)
print(h(10))
```

```python
def k(x):
    return x**2+3*x+2
print(k(2))
```

```python
#PART C
import numpy as np
import matplotlib.pyplot as plt

def plot_function(func,x_values,title):
    y_values=[]
    for x in x_values:
        try:
            y_values.append(func(x))
        except ValueError:
            y_values.append(None) # Skip invalid values
            plt.figure(figsize=(6, 4))
            plt.plot(x_values, y_values, label=title, color='orange')
            plt.axhline(0, color='black', linewidth=0.5)
            plt.axvline(0, color='black', linewidth=0.5)
            plt.legend()
            plt.grid()
            plt.title(title)
            plt.show()
            # Generate values for plotting
            x_values = np.linspace(-10, 10, 400) # Range for f(x) and k(x)
            x_values_positive = np.linspace(0.01, 10, 400) # For g(x) and h(x) (x > 0)
            # Plot each function separately
            plot_function(f, x_values, "f(x) = 1/x")
            plot_function(g, x_values_positive, "g(x) = sqrt(x)")
            plot_function(h, x_values_positive, "h(x) = ln(x)")
            plot_function(k, x_values, "k(x) = x^2 + 3x + 2")
```

```python
#Question13
def matrix_multiplication(matrix1,matrix2):
    rows1 = len(matrix1)
    cols1 = len(matrix1[0])
    rows2 = len(matrix2)
    cols2 = len(matrix2[0])
    if cols1 != rows2:
        return "Matrix cannot be multiplied: Incompatible diemensions."

    result = [[0 for _ in range(cols2)] for _ in range(rows1)]

    for i in range(rows1):
     for j in range(cols2):
        for k in range(cols1):
            result[i][j] += matrix1[i][k] * matrix2[k][j]

    return result
print(matrix_multiplication([[1,2],[3,4]],[[5,6],[7,8]]))
```

```python
#Question14
def decimal_to_binary(n):
    if n == 0:
        return "0"

    binary = ""
    while n > 0:
        remainder = n % 2
        binary = str(remainder) + binary
        n = n // 2

    return binary

print(decimal_to_binary(10))
```

```python
#Question15
def generate_prime_numbers(limit):
    primes = []
    for num in range(2,limit + 1):
        is_prime = True
        for i in range(2,int(num ** 0.5) + 1):
          if num % i == 0:
            is_prime = False
            break
        if is_prime:
            primes.append(num)
    return primes
print(generate_prime_numbers(10))
```
