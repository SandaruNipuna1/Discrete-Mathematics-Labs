```python
#Question9
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primes = [num for num in range(1, 21) if is_prime(num)]
print(", ".join(map(str, primes)))
```

```python
#Question10
def second_largest(numbers):
    unique_numbers = list(set(numbers))
    unique_numbers.sort(reverse=True)
    return unique_numbers[1] if len(unique_numbers) > 1 else None

numbers = [5, 8, 12, 6, 3, 9]
print("Second Largest:", second_largest(numbers))
```

```python
#Question12
def gcd(a, b):
    while b:
        a, b = b, a % b  #swap a with b and b with a % b
    return a

# Example input
num1, num2 = map(int, input("Enter two numbers separated by a comma: ").split(','))

# Compute and print GCD
print("GCD:", gcd(num1, num2))
```

```python
#Question13
import math

def is_strong_number(n):
    sum_of_factorials = sum(math.factorial(int(digit)) for digit in str(n))
    return sum_of_factorials == n

#getting input
num = int(input("Enter a number: "))

#Checking and printing the result
if is_strong_number(num):
    print("Strong Number")
else:
    print("Not a Strong Number")
```

```python
#Question14
def decimal_to_binary(n):
    binary = ""
    while n > 0:
        binary = str(n % 2) + binary #Get remainder and add to the left
        n //= 2
    return binary if binary else "0"

#getting input
num = int(input("Enter a decimal number: "))
#Convert and print the result
print("Binary:", decimal_to_binary(num))
```

```python
#Question15
import string

def is_pangram(sentence):
    sentence = sentence.lower()  #Convert to lowercase
    letters = set(sentence)  #Get unique characters
    return set(string.ascii_lowercase).issubset(letters)
# Example input
sentence = input("Enter a sentence: ")
# Checking and printing the result
if is_pangram(sentence):
    print("Pangram")
else:
    print("Not a Pangram")
```

```python
#Question3
def count_letters_digits(s):
    letters = sum(c.isalpha() for c in s)  # Count letters
    digits = sum(c.isdigit() for c in s)  # Count digits
    return letters, digits

# getting user input
text = input("Enter a string: ")

# Count and print result
letters, digits = count_letters_digits(text)
print(f"Letters: {letters}, Digits: {digits}")
```

```python
#Question9
# Getting the user input
num = int(input("Enter a number: "))

# Initialize sum and a list to store digits
sum_of_digits = 0
digits = []

temp = num 
while temp > 0:
    digit = temp % 10  # Get the last digit
    sum_of_digits += digit  # Add digit to sum
    digits.append(digit)  # Store digit for display
    temp //= 10  # Remove the last digit

# Reverse the list to show digits in correct order
digits.reverse()

# Print the result
print(" + ".join(map(str, digits)), "=", sum_of_digits)
```

```python
#Question11
# Function to find the maximum difference
def max_difference(lst):
    # Find the maximum and minimum values in the list
    max_val = max(lst)
    min_val = min(lst)
    
    # Return the difference between the maximum and minimum values
    return max_val - min_val

numbers = [10, 2, 8, 15, 1]

# Calculate and print the maximum difference
result = max_difference(numbers)
print(f"The maximum difference is: {result}")
```

```python
#Question12
# Function to calculate character frequencies
def char_frequency(string):
    freq_dict = {}  # Initialize an empty dictionary
    for char in string:
        # Update the frequency count for each character
        if char in freq_dict:
            freq_dict[char] += 1
        else:
            freq_dict[char] = 1
    return freq_dict

# Getting the user for input
user_input = input("Enter a string: ")

# Calculate and print the character frequencies
result = char_frequency(user_input)
print(result)
```

```python
#Question13
#Part a
# Function to compute the terms based on the recurrence relation
def compute_terms(n):
    # Initialization
    a = [2, 3]
    
    for i in range(2, n):
        next_term = a[i-1] + 2 * a[i-2]
        a.append(next_term)
    
    return a

# Number of terms to compute
n_terms = 10

# Compute the first 10 terms
sequence = compute_terms(n_terms)

# Print the first 10 terms
print("The first 10 terms of the sequence are:")
print(sequence)
print()
#Part b
# Calculate the sum of the first 10 terms
sum_of_terms = 0
for term in sequence:
    sum_of_terms += term

# Print the sum
print(f"The sum of the first 10 terms is: {sum_of_terms}")
```

```python
#Question14
#Part a
# Function to generate the first 15 terms of the arithmetic sequence
def generate_sequence(start, diff, n):
    sequence = []
    for i in range(n):
        term = start + i * diff
        sequence.append(term)
    return sequence

start_term = 5
difference = 4
num_terms = 15

sequence = generate_sequence(start_term, difference, num_terms)

# Print the first 15 terms
print("The first 15 terms of the sequence are:")
print(sequence)
print()

#Part b
# Calculate the sum of the first 15 terms
sum_of_terms = 0
for term in sequence:
    sum_of_terms += term

# Print the sum
print(f"The sum of the first 15 terms is: {sum_of_terms}")
print()

#Part c
# Formula to calculate the sum of the first n terms of an arithmetic sequence
def sum_arithmetic_sequence(a1, d, n):
    return n * (2 * a1 + (n - 1) * d) // 2

# Calculate the sum using the formula
sum_formula = sum_arithmetic_sequence(start_term, difference, num_terms)

# Print the result from the formula
print(f"The sum of the first 15 terms (calculated using the formula) is: {sum_formula}")
```

```python
#Question15
#Part a
#Part b
# Function to calculate population for each year
def calculate_population(initial_population, years):
    population = [initial_population]  # List to store population for each year
    for year in range(1, years):
        # Calculate the population for the current year
        if year > 1:
            current_population = 1.15 * population[year-1] + 20 - 5
        else:
            current_population = 1.15 * population[year-1] + 20
        population.append(current_population)
    return population

initial_population = 100
years = 10

# Calculate and print the population for each year
population_over_years = calculate_population(initial_population, years)

# Print population for each year
print("Year-wise population:")
for year, pop in enumerate(population_over_years):
    print(f"Year {year + 1}: {pop:.2f}")

#Part c
# Calculate the total population increase
total_increase = population_over_years[-1] - initial_population
print(f"\nTotal population increase over 10 years: {total_increase:.2f}")

#Part d
# Find the year when population first exceeds 250
for year, pop in enumerate(population_over_years):
    if pop > 250:
        print(f"\nThe population first exceeds 250 individuals in Year {year + 1}")
        break

#Part e
import matplotlib.pyplot as plt

# Create x (years) and y (population) arrays
x = list(range(1, years + 1))
y = population_over_years

# Plot the population trend
plt.plot(x, y)
plt.title("Population Growth Over Time")
plt.xlabel("Year")
plt.ylabel("Population")
plt.show()
```
