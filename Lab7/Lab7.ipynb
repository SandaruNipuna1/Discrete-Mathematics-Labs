```python
#PART A
#Question 1
#I
n = 1
lhs = sum(range(1, n+1))
rhs = n * (n + 1) // 2

print(f"n = {n}, LHS = {lhs}, RHS = {rhs}, Match: {lhs == rhs}")
```

```python
#II
def check_inductive_step(k):
    lhs_k = sum(range(1, k+1))
    rhs_k = k * (k + 1) // 2

    lhs_k_plus_1 = lhs_k + (k + 1)
    rhs_k_plus_1 = (k + 1) * (k + 2) // 2

    print(f"Assume P({k}): {lhs_k} == {rhs_k}")
    print(f"Prove P({k+1}): {lhs_k_plus_1} == {rhs_k_plus_1}")
    return lhs_k_plus_1 == rhs_k_plus_1

for k in range(1, 11):
    print(f"\nInductive Step from k = {k}")
    print(f"Valid: {check_inductive_step(k)}")
```

```python
#Question 3
#I
n = 1
lhs = sum([i**2 for i in range(1, n+1)])
rhs = n * (n+1) * (2 * n + 1) // 6

print(f"n = {n}, LHS = {lhs}, RHS = {rhs}, Match: {lhs == rhs}")
```

```python
#Question 4
import matplotlib.pyplot as plt
import math
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            return False
    return True

n_values = list(range(10, 1001, 10))
prime_counts = []
approximations = []

for n in n_values:
    primes = [i for i in range(2, n+1) if is_prime(i)]
    prime_counts.append(len(primes))
    approximations.append(n / math.log(n))

plt.plot(n_values, prime_counts, label="Actual π(n)", marker='o')
plt.plot(n_values, approximations, label="Approx. n/ln(n)", linestyle='--')
plt.xlabel('n')
plt.ylabel('Number of Primes <= n')
plt.title('Prime Number Distribution vs Prime Number Theorem')
plt.legend()
plt.grid(True)
plt.show
```

```python
#Question 5
import matplotlib.pyplot as plt
import math

def leibniz_pi(n_terms):
    pi_estimate = 0
    approximations = []
    errors = []
    
    for k in range(n_terms):
        term = ((-1)**k) / (2 * k + 1)
        pi_estimate += term
        pi_approx = 4 * pi_estimate
        approximations.append(pi_approx)
        errors.append(abs(math.pi - pi_approx))
        
    return approximations, errors
    
n_terms = 1000
approximations, errors = leibniz_pi(n_terms)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(range(1, n_terms + 1), approximations, label="π Approximation")
plt.axhline(math.pi, color='r', linestyle='--', label="Actual π")
plt.title("Approximation of π using Leibniz Series")
plt.xlabel("Number of Terms")
plt.ylabel("Approximate π")
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(range(1, n_terms + 1), errors, color='orange', label="Error")
plt.title("Error vs. Number of Terms")
plt.xlabel("Number of Terms")
plt.ylabel("Error (|π - Approx|)")
plt.yscale("log") # use log scale to show small error clearly
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
#Question 8
#I
import math

def binomial(n, r):
    return math.comb(n, r)

n = 1
lhs = sum([binomial(n, r) for r in range(0, n + 1)])
rhs = 2 ** n
print(f"n = {n}, LHS = {lhs}, RHS = {rhs}, Match: {lhs == rhs}")
```

```python
#II
def check_inductive_step(k):
    lhs_k = sum([binomial(k, r) for r in range(0, k + 1)])
    rhs_k = 2 ** k

    lhs_k_plus_1 = sum([binomial(k + 1, r) for r in range(0, k + 2)])
    rhs_k_plus_1 = 2 ** (k + 1)

    print(f"Assume P({k}): {lhs_k} == {rhs_k}")
    print(f"Prove P({k+1}): {lhs_k_plus_1} == {rhs_k_plus_1}")
    return lhs_k_plus_1 == rhs_k_plus_1

for k in range(1, 11):
    print(f"\nInductive Step from k = {k}")
    print(f"Valid: {check_inductive_step(k)}")
```

```python
#Question 9
#I
import math
import matplotlib.pyplot as plt

def approximate_e(n_terms):
    e_approx = sum([1 / math.factorial(k) for k in range(n_terms)])
    return e_approx

n_values = list(range(1, 31))
approximations = [approximate_e(n) for n in n_values]
actual_e = math.e
errors = [abs(actual_e - approx) for approx in approximations]
```

```python
#II
plt.figure(figsize=(10, 5))
plt.plot(n_values, approximations, label="Approximation of e", marker='o')
plt.axhline(y=actual_e, color='r', linestyle='--', label="Actual e (math.e)")
plt.xlabel("Number of Terms (n)")
plt.ylabel("Approximate Value of e")
plt.title("Approximation of Euler's Number e")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#III
plt.figure(figsize=(10, 5))
plt.plot(n_values, errors, label="Error", marker='x', color='orange')
plt.xlabel("Number of Terms (n)")
plt.ylabel("Absolute Error")
plt.title("Error in Approximating e vs. Number of Terms")
plt.yscale("log")  # Log scale to highlight how error decreases
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Question 10
#I
import matplotlib.pyplot as plt
import numpy as np
import math

n_max = 1000
harmonic_sums = []
ln_values = []

harmonic_sum = 0
for n in range(1, n_max + 1):
    harmonic_sum += 1 / n
    harmonic_sums.append(harmonic_sum)
    ln_values.append(math.log(n))
```

```python
#II
plt.figure(figsize=(10, 5))
plt.plot(range(1, n_max + 1), harmonic_sums, label="Harmonic Sum", color='blue')
```

```python
#III
plt.plot(range(1, n_max + 1), ln_values, label="ln(n)", color='red', linestyle='--')

plt.xlabel("n (Number of Terms)")
plt.ylabel("Sum")
plt.title("Harmonic Series vs ln(n)")
plt.legend()
plt.grid(True)
plt.show()
```
