```python
#Question 9
#(a)
from itertools import permutations

players = ["P1","P2","P3","P4","P5"]
pairs = list(permutations(players, 2))
print("All possible pairs:")
print(pairs)
```

```python
#(b)
pairs.sort()
print("The 4th pair in the sorted list of pairs:", pairs[3])
```

```python
#(c)
player_values = {"P1": 1, "P2": 2,"P3": 3, "P4": 4, "P5": 5}
count_odd_sum_pairs = 0
for pair in pairs:
    player1_value = player_values[pair[0]]
    player2_value = player_values[pair[1]]

    if(player1_value + player2_value) % 2 != 0:
        count_odd_sum_pairs += 1
print("Count of pairs where the sum of player number is odd:", count_odd_sum_pairs)
```

```python
#Question 11
#(a)
import itertools

students = ["S1","S2","S3","S4","S5","S6"]
pairs = list(itertools.combinations(students, 2))
print("All possible unique pairs:")
for pair in pairs:
    print(pair)
```

```python
#(b)
def is_odd(student):
    return int(student[1]) % 2 != 0

odd_pairs = [pair for pair in pairs if is_odd(pair[0]) and is_odd(pair[1])]
print(odd_pairs)
```

```python
#(c)
s1_pairs = [pair for pair in pairs if "S1" in pair]
print("Total pairs where one student is 'S1':", len(s1_pairs))
```

```python
#Question 12
#(a)
from itertools import product

chars = ["X","Y","Z"]
codes = list(product(chars, repeat=3))
print(codes)
```

```python
#(b)
matching_codes = ["".join(code) for code in codes if code[0] == code[2]]
print("Codes that start and end with the same character:")
for code in matching_codes:
    print(code)
```

```python
#(c)
valid_codes = ["".join(code) for code in codes if code[0] != code[1] and code[1] != code[2]]
print("Codes where no two consecutive characters are the same:")
print(len(valid_codes))
```

```python
#Question 15
#(a)
from itertools import permutations

gems = ["Ruby", "Ruby", "Sapphire", "Sapphire", "Emerald", "Emerald"]
unique_arrangements = set(permutations(gems))
print("All possible arrangements:")
for arrangement in unique_arrangements:
    print(arrangement)
```

```python
#(b)
arrangements_without_rubies_together = [
    arrangement for arrangement in unique_arrangements
    if "Ruby" not in arrangement or arrangement.index("Ruby") != arrangement.index("Ruby", arrangement.index("Ruby") + 1)
]
print("Arrangements where the two rubies are not next to each other:")
for arrangement in arrangements_without_rubies_together:
    print(arrangement)
```

```python
#(c)
arrangements_with_unique_first_3 = [
    arrangement for arrangement in unique_arrangements
    if set(arrangement[:3]) == {"Ruby","Sapphire","Emerald"}
]
print("Arrangements where the first 3 positions contain one Ruby, one Sapphire, and one Emerald:")
for arrangement in arrangements_with_unique_first_3:
    print(arrangement)

print("Number of such arrangements:", len(arrangements_with_unique_first_3))
```
