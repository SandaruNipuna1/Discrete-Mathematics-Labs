```python
#Question5
sale_event_a = {1001, 1002, 1003, 1004, 1005, 1006}
sale_event_b = {1004, 1005, 1006, 1007, 1008, 1009}
#(a)
#customers who participated in both sales
both = sale_event_a.intersection(sale_event_b)
print(both)
#(b)
#customers who participated only in Sale Event A
onlyA = sale_event_a.difference(sale_event_b)
print(onlyA)
#(c)
#customers who participated only in Sale Event B
onlyB = sale_event_b.difference(sale_event_a)
print(onlyB)
#(d)
#customers who participated in at least one sale
atleastOne = sale_event_a.union(sale_event_b)
print(atleastOne)
#(e)
#customers who participated in exactly one sale
exactlyOne = sale_event_a.symmetric_difference(sale_event_b)
print(exactlyOne)
#(f)
#customers who did not participate in both sales simultaneously
count = len(exactlyOne)
print(count)
#new set
new_customers = {1008, 1009, 1010, 1011}
#(g)
#Customers who are new
new = new_customers.difference(atleastOne)
print(new)
#(h)
#Customers who have participated in at least one past sale and are now included in the new customers list
nowNew = atleastOne.intersection(new_customers)
print(nowNew)
```

```python
#Question14
#a
scores = [78,85,95,62,88,75,95,85,62,90]
unique_scores = list(dict.fromkeys(scores))
print("Unique scores: ", unique_scores)
```

```python
#b
average_score = round(sum(scores)/len(scores), 2)
print("Average score:", format(average_score, " .2f"))
```

```python
#c
top_3_scores = sorted(set(scores), reverse=True)[:3]
print("Top 3 highest unique scores:", top_3_scores)
```

```python
#d
score_counts = {score: scores.count(score) for score in set(scores)}
print("Score frequency dictionary:", score_counts)
```

```python
#e
updated_scores = ["Fail" if score < 70 else score for score in scores]
print("Updated scores list: ", updated_scores)
```

```python
#f
passing_scores = [score for score in scores if score >= 70]
print("Passing scores:", passing_scores)
```

```python
#g
accending_passing_scores = sorted(passing_scores)
print("Passing scores in ascending order", accending_passing_scores)
```

```python
#h
print("The original list: ", scores)
print("Passing scores list: ", passing_scores)
```

```python
#i
unique_sorted_scores = sorted(unique_scores)
n = len(unique_sorted_scores)
if n % 2 == 1;
median = unique_sorted_scores[n // 2]
else:
median = (unique_sorted_scores[n // 2 -1] + unique_sorted_scores[n // 2]) / 2
print(f"Median score: {median}")
```
