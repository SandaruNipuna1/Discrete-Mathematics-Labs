```python
#Question2
from itertools import product
char = ["A","B","C","D","E"]
codes = list(product(char,repeat=3))
print("All codes of 3 characters:")
print(codes)
```

```python
A_codes = []
for x in codes:
    if "A" in "".join(x):
        A_codes.append(x)
print("Number of codes include the letter A at least once:")
print(len(A_codes))
print("Codes include the letter A at least once:")
print(A_codes)
```

```python
uni_char_codes = []
for x in codes:
    if len(set(x)) > 1:
        uni_char_codes.append(x)
print(uni_char_codes)
for x in uni_char_codes:
    if list(x) == sorted(x):
        print(x)
```

```python
#Question4
from itertools import combinations_with_replacement as cmr
biscuits = ["B1","B2","B3","B4"]
sets = list(cmr(biscuits,6))
print("Different sets:")
print(sets)
print("Number of different sets:")
print(len(sets))
```

```python
at_least_fla = []
for x in sets:
    if len(set(x)) == 4:
        at_least_fla.append(x)
print(len(at_least_fla))
```

```python
from itertools import combinations
outof3_4 = list(combinations(biscuits,3))
print(outof3_4)
```

```python
from itertools import combinations_with_replacement
bis6 = []
for x in outof3_4:
    cwr = list(combinations_with_replacement(x,6))
    for v in cwr:
        bis6.append(v)
print(len(set(bis6)))
```

```python
#Question8
from itertools import permutations

people = [1, 2, 3, 4, 5, 6]
all_arrangements = permutations(people)

valid_arrangements = []
for seating in all_arrangements:
    if all(abs(seating[i] - seating[i+1]) > 1 for i in range(len(seating) - 1)):
        valid_arrangements.append(seating)

print("Total valid seating arrangements:")
print(valid_arrangements)
```

```python
#Question9
from itertools import combinations_with_replacement

flowers = ["Rose", "Lily", "Daisy"]

all_bouquets = combinations_with_replacement(flowers, 5)

valid_bouquets = [bouquet for bouquet in all_bouquets if bouquet.count("Rose") >= 3]

print("Total bouquets with at least 3 Roses:", len(valid_bouquets))
print(valid_bouquets)
```

```python
#Question12
from itertools import permutations

digits = [1, 2, 3, 4, 5, 6, 7, 8, 9]

all_grids = permutations(digits)

valid_grids = []
for grid in all_grids:
    row1, row2, row3 = grid[:3], grid[3:6], grid[6:9]

    col1 = row1[0] + row2[0] + row3[0]
    col2 = row1[1] + row2[1] + row3[1]
    col3 = row1[2] + row2[2] + row3[2]

    diag1 = row1[0] + row2[1] + row3[2]
    diag2 = row1[2] + row2[1] + row3[0]  

    if (
        sum(row1) == sum(row2) == sum(row3) == 15 and
        col1 == col2 == col3 == 15 and
        diag1 == diag2 == 15
    ):
        valid_grids.append((row1, row2, row3))

print("Total valid 3*3 grids:", len(valid_grids))
for grid in valid_grids:
    for row in grid:
        print(row)
    print("-" * 10)
```

```python
#Question13
from itertools import permutations

objects = ['A','B','C','D']
block = ['AB','BA']
remaining_objects = ['C','D']

valid_arrangements = []
for block_arrangements in block:
    arrangements = permutations([block_arrangements] + remaining_objects)
    valid_arrangements.extend(arrangements)
print("Total valid arrangements:", len(valid_arrangements))
for arrangement in valid_arrangements:
    print(arrangement)
```

```python
#Question15
from itertools import product

characters = ["X", "Y", "Z"]

all_words = product(characters, repeat=5)

valid_words = [''.join(word) for word in all_words if all(word[i] != word[i+1] for i in range(4))]

print("Total valid 5-letter words:", len(valid_words))
print(valid_words)
```
